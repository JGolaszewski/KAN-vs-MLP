## Kolmogorov-Arnold representation theorem

The theorem states that every multivariate continuous function $ f : [0,1]^n \to \mathbb{R} $, where **n** is a number of variables, can be represented as a superstition of continous single-variable functions 

$$
f(x_1, x_2, \ldots, x_n)
= \sum_{q=1}^{2n + 1} \Phi_q \left( \sum_{p=1}^{n} \psi_{pq}(x_p) \right)
$$
where  
$ \Phi_q :\mathbb{R}\to \mathbb{R}$  
$ \psi_{pq}: [0,1] \to \mathbb{R} $  
$ x_p \in [0,1]$  
n = number of input neurons 

But why is it usefull in the field of neural Network?  
To understand its importance lets have a look at the basic neural network model

# Traditional neural network architecture 

In a traditional approach a model consists of 3 layers:
1. Input layer. 
2. Hidden layer
3. Output  

Let's further examine what each layer does within the model

## Input layer

It's a foundation of the every neural network model. It provides data on which basis the final output will be computed. It also can be interpreted as a set of features necessary for the model to give a viable answer to the problem.  
As analogyto decide whether an animal is a Zebra you need an animal shape and a information about its pattern. Those information can be viewed as input features. What's important, those features are **INDEPENDENT** from eachother from the models perspective. 

## Hidden layer

Hidden layer is the core of the neural network. It consists of wages and biases allowing to transform numerical input into output. Through training the state(values of wages and bias) of the hidden layer is adjusted to the given problem in roder to give acurate prediction. In a Zera case hidden layer would be all neural procces on which basis we decide that the animal is a Zebra. For example if the animal is in the hippo shape and has gray colour, our brain can combine and transform those informations to predict that its very unlikely for this animal to be a zebra. This process is exactly what hidden layer does. Effectively it constructs a pattern represented by mathematical formula on the basis of which the final decision is beeing made. More thorough explenation of neural network compution is given in traditional neural network document. 

## Output layer


Output layer is the icing on the cake. It's the last step toward gaining a meaningful answer from the input features. The output layer, unlike hidden layer, doesn't take part in a thought process. Instead it changes the values from hidden layer to a form that is desireable. In a zebra example it would be transforming final neural informations into "NO"-It can be a zebra since it's a hippo. The final transformation can be done in a multiple ways-it all depends on what kind of answer you want for a problem. It can be number, boolean information or even a text written by language model

# Kan network architecture-theoretical

## Introduction 

Since both **Input** and **Output** layers are identycal in structure as in the classic MLP network, in this chapter we will closer examine how hidden layer works and discuss how implementation of KAN network may look like. One thing worth to note however, is the fact that all input values must be within the range of [0,1]

## Hidden layer

Unlike the hidden layer in the classic MLP network, the KAN hidden layer can be represented best as single row of neurons, each connected by one node with the all input values. Every node has its own function $\psi(x)$ that changes value of the x input. Next, all the $\psi(x_p)$ connected with the neuron are added and then transformed using the $\Phi$ function to get the value of the neuron.  $$ neuron_q = \Phi_q \left( \sum_{p=1}^{n} \psi_{pq}(x_p) \right)$$
 The same mathematical operations are used for all the neurons in the hidden layer. Also the number of neurons in the layer should not be random. The theorem states that if **n** is the number of innput neurons, than you need at least 2n+1 neurons in the hidden layer in order to fully map a function(relationship between input and output we want to find).

## Output layer 

Output layer in the KAN network consists of only one otput neureon. Its value is just a sum of all neurons from the hidden layer. $$ Output = \sum_{q=1}^{2n + 1} \Phi_q \left( \sum_{p=1}^{n} \psi_{pq}(x_p) \right)$$

## Limitations

According to the theory $\Phi$ and $\psi$ functions have no restrictions on how complicated they are, as long as they map values accordingly($ \Phi :\mathbb{R}\to \mathbb{R}$  and $ \psi: [0,1] \to \mathbb{R} $ ). It is easy to imagine that some functions are very difficult to compute, so in order to make implementation more practical we have to consider what are the most usefull forms of the functions. Ideally they should be relatively easy to compute and allow huge flexibility to demonstrate complex patterns between data. Also the functions have to be differentiable, smooth and numerically stable in order to allow effective training of such network.  

# Actual implementation

In this paragraph we will duscuss more detailed implementation of the KAN network including functions to overcome chalanges and limitations of the practical implementation. 

## Functions 

According to the theory functions do the following mapping:  
$\psi: [0,1] \to \mathbb{R}$  
$ \Phi :\mathbb{R}\to \mathbb{R}$   
There are many possibilities how to realise those functions. The most popular approach is to use **B-splines**
Before we discuss the usage of B-splines we should further examine properties of KAN functions. In reality, they can be not-smooth making the network impossible to learn. Also, just 2 layers, as in the theory, may not be enough for network to learn complex functions. Thats why we have to rethink the design of the network and try to find diffrent aproach in its implementation. Somehow we have to find a way to generalize a network and find a way to create more layers. But how can we represent a layer in the KAN network?

## KAN layers

It turns out that every KAN layer consisting of $n_{in}$ dimensional inputs and $n_{out}$ dimensional outputs can be represented as matrix of 1D functions $ \Phi = \{\psi_{q,p}\}$, where: $ p = 1, 2,...,n_{in}$ and $q = 1,2,...,n_{out} $  
Due to the Kolmogorov-Arnold representation theorem the values of $n_{in}$ and $n_{out}$ are equal to:  
$n_{in} = n$, $n_{out} = 2n + 1$ for inner layer   
$n_{in} = 2n + 1$, $n_{out} = 1$ for outer(last) layer  

By doing so we can represent shape of the whole KAN network as $[n_0, n_1,...,n_L]$ where $n_i$ is the number of nodes in the $i^{th}$ layer of the network. We can also denote a single neuron as a pair of $(l,i)$ values where:     
-$l$ is number of the layer   
-$i$ is the number of the neuron in that layer  
According to the theory, it means that between every $l$ and $l+1$ layer there are $n_ln_{l+1}$ activation functions.   
It allows us to denote activation function between $(l,i)$ and $(l+1,i)$ as:   
$\psi_{l,j,i}$ where $l = 0,...,L-1$,  $i = 1,...,n_l$,  $j = 1,...,n_{l+1}$

The pre-activation value(argument of the $\psi$ function) is $x_{l,i}$ whereas post-activation value will be denoted as $\tilde{x}_{l,j,i}$.  
$\tilde{x_{l,j,i}}$ $\equiv$ $\psi_{l,j,i}(x_{l,i})$
It means that the activation value of the $(l+1,j)$ is the sum of the all incoming post-activation in the layer $l$. 
$$x_{l+1,j} = \sum_{i=1}^{n_l}\tilde{x}_{l,j,i} = \sum_{i=1}^{n_l}\psi_{l,j,i}(x_{l,i})$$
In the matrix form: 

$$
x_{l+1} =
\underbrace{
\begin{pmatrix}
\psi_{l,1,1}(\cdot) & \psi_{l,1,2}(\cdot) & \dots &\psi_{l,1,3}(\cdot) \\
\psi_{l,2,1}(\cdot) & \psi_{l,2,2}(\cdot) &\dots &\psi_{l,2,3}(\cdot)\\
\vdots & \vdots &  & \vdots\\
\psi_{l,n_{l+1},1}(\cdot)&\psi_{l,n_{l+1},2}(\cdot)& &\psi_{l,n_{l+1},n_l}(\cdot)
\end{pmatrix}
}_{\Phi_l}
x_l
$$
Each $\Phi_l$ is corresponding to the $l^{th}$ layer of the KAN network.  
Now we can clearly see that KAN network is indeed a composition of L layers, similairly to standard MLP network. 
If we consider input vector $x_0 \in R^{n_0}$, the output would be: 
$$KAN(x) = (\Phi_{L-1}\circ\Phi_{L-2}\circ\dots\circ\Phi_1\circ\Phi_0)x$$
$$MLP(x) = (W_{L-1}\circ\sigma\circ W_{L-2}\circ\sigma\dots W_{1}\circ\sigma\circ W_{0})x$$
Morover we can rewrite our KAN equation. Assuming that output dimension is 1(jus a number) we can define our KAN function $f(x) \equiv KAN(x)$ 
$$
f(x) = 
\sum_{i_{L-1} = 1}^{n_{L-1}} 
\psi_{L-1, i_L, i_{L-1}} 
\Biggl(
  \sum_{i_{L-2}=1}^{n_{L-2}} 
  \dots
    \Biggl(
      \sum_{i_2=1}^{n_2} 
      \psi_{2, i_3, i_2} 
      \Biggl(
        \sum_{i_1=1}^{n_1} 
        \psi_{1, i_2, i_1} 
        \Biggl(
          \sum_{i_0=1}^{n_0} 
          \psi_{0, i_1, i_0}(x_{i_0})
        \Biggr)
      \Biggr)
    \Biggr)
\Biggr)
$$


As we can see, all operations are differentiable so it is possible to train KAN network using backpropagation, just like a regular MLP network.

# Parameter count 

In order to count parameters let us assume the following network:   
1. $L$ = depth of the network  
2. Layers are of equal size: $n_0 = n_1 =...= n_L = N$   
3. Each spline of order k on G intervals (for G + 1 grid points, k = 3 usually)  

There are $O(N^2 L(G+K))$~$O(N^2 LG)$ parameters in total 
In contrast a MLP network with depth L and N only needs $O(N^2L)$ parameters. 
At first it may appear as more efficient, in reality however, KAN usually requires much smaller number of N parameters(layer size) than regular MLP. Morove KAN, through its structure, is able to provide much better generalisation along with interpretability. For instance, if we take N=L=1 we receive splain approximation, which already is able to provide us useful informations.
